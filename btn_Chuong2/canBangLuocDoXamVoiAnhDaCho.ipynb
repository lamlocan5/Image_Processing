```python
import cv2
path_in  = "Fig0310(b)(washed_out_pollen_image)100kb.jpg"
path_out = "Fig0310(b)(washed_out_pollen_image)100kb_OUT.jpg"

```

```python
gray = cv2.imread(path_in, cv2.IMREAD_GRAYSCALE)
gray
```

```python
equalized = cv2.equalizeHist(gray)
equalized
```

```python
cv2.imwrite(path_out, equalized)
```

```python
import matplotlib.pyplot as plt

# Hiển thị trong notebook
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.title("Original")
plt.imshow(gray, cmap="gray")
plt.axis("off")

plt.subplot(1,2,2)
plt.title("Equalized")
plt.imshow(equalized, cmap="gray")
plt.axis("off")

plt.show()
```
